# 项目：评估和清理英国电商公司销售数据

## 分析目标

此数据分析的目的是，根据市场销售数据，挖掘畅销产品，以便制定更有效的市场策略来提升营收。

本实战项目的目的在于练习评估数据干净和整洁度，并且基于评估结果，对数据进行清洗，从而得到可供下一步分析的数据。

## 简介

原始数据集记录了一家英国在线零售公司在2010年12月1日至2011年12月9日期间的所有交易情况，涵盖了该公司在全球不同国家和地区的业务数据。该公司主要销售覆盖各个场景的礼品，包括但不限于生日礼品、结婚纪念品、圣诞礼品等等。该公司的客户群体主要包括批发商和个人消费者，其中批发商占据了相当大的比例。

数据每列的含义如下：
- `InvoiceNo`: 发票号码。6位数，作为交易的唯一标识符。如果这个代码以字母“c”开头，表示这笔交易被取消。
- `StockCode`: 产品代码。5位数，作为产品的唯一标识符。
- `Description`: 产品名称。
- `Quantity`: 产品在交易中的数量。
- `InvoiceDate`: 发票日期和时间。交易发生的日期和时间。
- `UnitPrice`: 单价。价格单位为英镑（£）。
- `CustomerID`: 客户编号。5位数，作为客户的唯一标识符。
- `Country`: 国家名称。客户所居住的国家的名称。

## 清理数据

```python
cd = od.copy()
```

将InvoiceDate从str数据转换为datatime数据

```python
cd["InvoiceDate"] = pd.to_datetime(cd["InvoiceDate"])
cd["InvoiceDate"]
```

将CustomerID从float转换成为str数据并且只保留整数部分

```python
cd["CustomerID"] = cd["CustomerID"].astype(str)
cd["CustomerID"]
```

```python
cd["CustomerID"] = cd["CustomerID"].str.slice(0,-1)
cd["CustomerID"]
```

!!做删减数据的时候，最好的都后面使用的是负数，而不要使用正数去推。因为有时候他的客户编号不一定只有5位数，他可能会有6位数

删除description缺失的观察值

```python
cd = cd.dropna(subset = ["Description"])
cd["Description"]
```

将英国和美国的不同的表达都改为 United Kingdom and United States

```python
cd["Country"] = cd["Country"].replace({"UK":"United Kingdom","U.K.":"United Kingdom","USA":"United States"})
cd["Country"].value_counts()
```

```python
len(cd[cd["Country"]=="U.K."])
```

把quantity为负数的观察值清洗掉,并检查负数的值是否还存在

```python
cd = cd[cd["Quantity"]>= 0]
len(cd[cd["Quantity"]< 0])
```

```python
#把单价为负数的也给清理掉c
cd = cd[cd["UnitPrice"]>=0]
len(cd[cd["UnitPrice"]<0])
```

## 保存清理后的数据

```python
cd.to_csv("e_commerce_clean")
```
